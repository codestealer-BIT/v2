```python
import jsonlines
import pickle
from tqdm import tqdm

data_dict = {}
with jsonlines.open('/mnt/bn/jiny-ttls-i18n-fr1q/ttls_hive_data/tt_captions_for_music.jsonl', 'r') as reader:
    for item in tqdm(reader):
        item_id = item['item_id']
        if item_id in data_dict:
            if data_dict[item_id] == '':
                data_dict[item_id] = item['video_caption_v2']
        else:
            data_dict[item_id] = item['video_caption_v2']

print(len(data_dict))
```

```python
cnt = 0
tt_cnt = 0
for item_id in tqdm(data_dict):
    if data_dict[item_id] == '':
        cnt += 1
    else:
        if 'Negative: ' in data_dict[item_id]:
            data_dict[item_id] = data_dict[item_id].replace('Negative: ', '')
        elif 'Positive: ' in data_dict[item_id]:
            data_dict[item_id] = data_dict[item_id].replace('Positive: ', '')
    
print(cnt, tt_cnt) 
```

```python
with open('/mnt/bn/jiny-ttls-i18n-fr1q/ttls_hive_data/tt_captions_for_music.pkl', 'wb') as fw:
    pickle.dump(data_dict, fw)

# Negative: 
# Positive: 
```

```python
import torch
import os
import pandas as pd


model_v1_result = '/mnt/bn/jiny-ttls-i18n-fr1q/results/video_music_clip_dedup_v3_large_queue/checkpoint-200000/ema_custom/music_library_top1000_dict.pt'
model_v1_result = torch.load(model_v1_result)
model_v2_result = '/mnt/bn/jiny-ttls-i18n-fr1q/results/video_music_text_clip/checkpoint-200000/ema_custom/music_library_top1000_dict.pt'
model_v2_result = torch.load(model_v2_result)
```

```python
output_data = {'video_id': [], 'top20_music_id': []}

for video_id in model_v1_result:
    output_data['video_id'].append(str(video_id))
    top50_music_id = [str(music_id) for music_id in model_v1_result[video_id][:20]]
    output_data['top20_music_id'].append('\n'.join(top50_music_id))

output_data = pd.DataFrame(output_data)
output_data.to_excel('model_v1_result.xlsx', index=False)
```
